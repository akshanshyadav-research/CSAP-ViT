# CSAP-ViT quickstart

This notebook demonstrates token pruning with synthetic tensors. For real evaluations, use the saved configs and commands in the README. No dataset or weights are bundled.

````python
from pathlib import Path
import sys
root = Path.cwd() if (Path.cwd() / "similarity_pruning").exists() else Path.cwd().parent
sys.path.insert(0, str(root))
import torch
from similarity_pruning import prune_tokens
````

````python
tokens = torch.tensor([[[1., 0.], [1., 0.], [0., 1.], [-1., 0.]]])
print("Input:", tokens)
print("Drop most similar patch:", prune_tokens(tokens, drop_rate=1/3))
print("Drop least similar patch:", prune_tokens(tokens, drop_rate=1/3, drop_most_similar=False))
````
